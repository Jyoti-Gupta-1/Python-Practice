```python
#aggregate functions

import numpy as np
a=np.array([20,30,40,50])

print(np.sum(a))
print(np.min(a))
print(np.max(a))
print(np.size(a))  # size is used for count
print(np. mean(a))
print(np.cumsum(a)) #cumulative sum
print(np.cumprod(a))  #cumulative product
```

```python
a=[100,150,199,200,250,130]
b=[10,50,30,40,30,10]

price=np.array(a)
quantity=np.array(b)

print (price,"\n", quantity)

c= np.cumprod([price,quantity],axis=0) #if there are two columns then axis should be provided
print(price*quantity)   #both gives same results

print(c[1].sum()) #gives sum of cumulative products

```

```python
#statstical functions
import statistics as stats
baked_food=[200,300,150,130,200,280,170,188]
a=np.array(baked_food)
print(np.mean(baked_food))
print(np.median(baked_food))  #central value after sorting
print(stats.mode(baked_food))  #mode is most occuring number . for mode we import statistics
print(np.std(baked_food))  #standard deviation . lower STD is better and product is good,
 # it tells how far the values are from the mean value, widely used to tell the spread of data
print(np.var(baked_food))  #var is nothing but square of standard deviation

```

```python
tobacco_consumption=[30,50,10,30,50,40]
deaths=[100,120,70,100,120,112]
# -1 represents inversely proportional relationship
# 1 represents proportional relationship
# 0 means no relationship
print(np.corrcoef([tobacco_consumption,deaths]))  # here if one is incresing the other one is also increasing
```
